```python
from pathlib import Path
```

```python
from anomalib.data import Folder
```

```python
from anomalib.engine import Engine
from anomalib.models import Patchcore
```

```python
datamodule = Folder(
    name="hackathon",
    root=Path("../../datasets/hackathon"),
    normal_dir="train/good",  # Subfolder containing normal images
    abnormal_dir="test/bad",
    train_batch_size=32,
    eval_batch_size=32
)
```

```python
pre_processor = Patchcore.configure_pre_processor(image_size=(256, 256))
```

```python
model = Patchcore(
    backbone="resnet18",  # Feature extraction backbone
    layers=["layer2", "layer3"],  # Layers to extract features from
    pre_trained=True,  # Use pretrained weights
    num_neighbors=9,  # Number of nearest neighbors,
    pre_processor=pre_processor
)
```

```python
engine = Engine(
    max_epochs=1,  # Patchcore typically needs only one epoch
    accelerator="auto",  # Automatically detect GPU/CPU
    devices=1,  # Number of devices to use
)
```

```python
engine.fit(
    model=model,
    datamodule=datamodule,
)
```

```python
datamodule_test = Folder(
    name="hackathon",
    root=Path("../../datasets/hackathon"),
    normal_dir="test/good",  # Subfolder containing normal images
    abnormal_dir="test/bad",
    train_batch_size=32,
    eval_batch_size=32
)
```

```python
engine.test(model=model, datamodule=datamodule_test)
```
