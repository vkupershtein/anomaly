```python
import os
import cv2
import albumentations as A
from glob import glob
```

```python
input_dir = "input_bad"
output_dir = "augmented_bad"
os.makedirs(output_dir, exist_ok=True)
```

```python
def center_crop_resize(image, size=256):
    h, w = image.shape[:2]
    min_dim = min(h, w)
    
    # Center coordinates
    start_x = (w - min_dim) // 2
    start_y = (h - min_dim) // 2
    cropped = image[start_y:start_y+min_dim, start_x:start_x+min_dim]
    
    resized = cv2.resize(cropped, (size, size))
    return resized
```

```python
transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.2),
    A.RandomRotate90(p=0.5),    
    A.RandomBrightnessContrast(p=0.5)
])
```

```python
images = glob(os.path.join(input_dir, "*.jpeg")) + glob(os.path.join(input_dir, "*.png"))
```

```python
n_target = 20
counter = 0

# === 4. Generation of augmented copies ===
while counter < n_target:
    for img_path in images:
        image = cv2.imread(img_path)
        if image is None:
            continue

        # Crop + resize
        image = center_crop_resize(image, size=256)

        # Augmentation
        augmented = transform(image=image)["image"]

        out_path = os.path.join(output_dir, f"aug_{counter:03d}.png")
        cv2.imwrite(out_path, augmented)

        counter += 1
        if counter >= n_target:
            break
```
