```python
from pathlib import Path
import cv2
import numpy as np
import torch
```

```python
ckpt_path = "results/Patchcore/hackathon/latest/weights/lightning/model.ckpt"
```

```python
image_path = "augmented_good/aug_002.png"
```

```python
from anomalib.engine import Engine
```

```python
from anomalib.models import Patchcore
```

```python
pre_processor = Patchcore.configure_pre_processor(image_size=(256, 256))
```

```python
model = Patchcore(
    backbone="resnet18",  # Feature extraction backbone
    layers=["layer2", "layer3"],  # Layers to extract features from
    pre_trained=True,  # Use pretrained weights
    num_neighbors=9,  # Number of nearest neighbors,
    pre_processor=pre_processor
)
```

```python
engine = Engine()
```

```python
result = engine.predict(model=model, ckpt_path=ckpt_path, data_path=image_path)
```
